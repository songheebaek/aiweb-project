```python
# 실습 준비 — 03주차 1차원 데이터 정리 리마인드 · 퀴즈 · LaTeX
# 이 셀을 먼저 한 번 실행하세요. 데이터가 없으면 아래 셀들이 전부 실패합니다.
import os, pathlib, urllib.request

BASE = "https://raw.githubusercontent.com/aprilslab/statistics-lab/main/data/"
FILES = []

pathlib.Path("data").mkdir(exist_ok=True)
for name in FILES:
    for dest in (pathlib.Path(name), pathlib.Path("data") / name):
        if not dest.exists():
            urllib.request.urlretrieve(BASE + name, dest)

# '../data/x.csv' 로 읽는 노트북 대응 — 상위 폴더에도 같은 data/ 를 걸어둔다.
# 절대경로(/data)로 박으면 cwd 가 /content 가 아닐 때 깨지므로 상대경로로 건다.
try:
    parent = pathlib.Path("..") / "data"
    if not parent.exists():
        os.symlink(pathlib.Path("data").resolve(), parent)
except OSError:
    pass

print("준비 완료:", ", ".join(FILES) if FILES else "(내려받을 데이터 없음)")
```

## 수식을 쓰는 세 가지 방법

같은 LaTeX 을 어디에 어떻게 감싸느냐에 따라 쓰는 법이 다르다. **안에 들어가는 LaTeX 본체는 셋 다 같다.**

| 방법 | 어디에 쓰나 | 화면에서 |
|---|---|---|
| `$ 수식 $` | 마크다운 셀 · 문장 속에 섞어 | 글줄 안에 그대로 |
| `$$ 수식 $$` | 마크다운 셀 · 문단으로 떼어 | 가운데 정렬, 크게 |
| `display(Math(r'수식'))` | 코드 셀 · 실행 결과로 | `Out[n]` 으로 나옴 |

아래 두 셀이 같은 수식을 서로 다른 방법으로 낸다. **이 셀을 더블클릭하면 마크다운 원본을 볼 수 있다.**

문장 속에 넣으면 이렇게 된다 — 평균은 $\bar{x}$ 로 적는다.

문단으로 떼면 이렇게 된다.

$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i
$$

### r'' 을 빼면 깨진다

코드 셀에서 `Math()` 를 쓸 때는 따옴표 앞에 `r` 을 붙인다. 파이썬이 따옴표 안의 백슬래시를 먼저 해석하기 때문이다.

| 쓴 것 | raw string 일 때 | 그냥 따옴표일 때 |
|---|---|---|
| `\frac` | `\frac` 그대로 | 폼피드 문자 + `rac` |
| `\nu` | `\nu` 그대로 (그리스 문자 뉴) | 줄바꿈 + `u` |
| `\times` | `\times` 그대로 | 탭 + `imes` |

마크다운 셀의 `$…$` 에는 이 문제가 없다. 파이썬 문자열이 아니기 때문이다.

```python
# 위 마크다운 셀과 같은 수식을 코드로 낸다
display(Math(r'\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i'))

# r 을 빼면 어떻게 되는지 직접 확인해 보자
display(Math(r'\frac{1}{n}'))   # 분수가 나온다
display(Math('\frac{1}{n}'))    # 깨지거나 엉뚱한 것이 나온다
```

## 02. 1차원 데이터 정리


### 평균값 ( mean, 기댓값, E(X), $\mu$ )
$$
\bar{x} = \frac{1}{n}{\sum_{i=1}^{n} x_i}
$$

* `\bar{x}` -> $\bar{x}$
* `\frac{1}{n}` -> $\frac{1}{n}$
* `x_i` -> x_i
* `\sum_{i=1}^{n} x_i` -> $\sum_{i=1}^{n} x_i$



```python
display(Math(r''))
```

### 분산

$$
{S}^2 = \frac{1}{n}{\sum_{i=1}^{n} ({x_i - \bar{x}})^2}
$$

* `S^2` -> $S^2$
* `({x_i - \bar{x}})^2` -> $({x_i - \bar{x}})^2$

```python
display(Math(r''))
```

### 표준편차
<center>
$
S^2 = \sqrt{S^2} = \sqrt{\displaystyle \frac{1}{n} \sum_{i=1}^n (x_i - \bar{x})^2}
$
</center>

* `\sqrt{S^2}` -> $\sqrt{S^2}$

```python
display(Math(r''))
```

### 범위 (range)

$$
{Rg} = {x_{\max} - x_{\min}}
$$

* `x_{\max}` -> $x_{\max}$
* `x_{\min}` -> $x_{\min}$

```python
display(Math(r''))
```

### 표준화 (standardization)

$$
{z_i} = \frac{{x_i - \bar{x}}}{{S}}
$$

```python
display(Math(r''))
```

### 편차값

## 03. 2차원 데이터 정리

### 공분산

$$
S_{xy} = \frac{1}{n}{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}
$$

```python
display(Math(r''))
```

### 상관계수

$$
r_{xy} = \frac{1}{n}\sum_{i=1}^{n}(\frac{x_i - \bar{x}}{S_x})(\frac{y_i - \bar{y}}{S_y}) = \frac{S_{xy}}{{S_x}{S_y}}
$$

```python
display(Math(r''))
```
